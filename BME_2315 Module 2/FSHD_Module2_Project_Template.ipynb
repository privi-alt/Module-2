## Disease Background: 

Symptoms: 
Facioscapulohumeral muscular dystrophy (FSHD) is a progressive genetic disorder where severity and progression varies depending on the individual. Common symptoms include weaknesses in the face, shoulder, hip, lower leg and upper arm. Facial weaknesses appear early on, making movements like smiling, whistling and closing the eyes difficult. Weakness around the shoulder blades is an indicator of FSHD, resulting in protruded shoulder blades and causing difficulty while performing regular activities with the shoulders. Other muscular areas can affect balance, mobility due to foot drop, and seating, creating an uncomfortable quality of life for individuals living with FSHD. Roughly 20% of people living with FSHD use a wheelchair to help with mobility. Other side effects that accompany FSHD conditions include respiratory complications, blindness, and dysphagia. While symptoms vary widely among individuals, most develop symptoms from their teens, and currently there is no established cure for FSHD.(Muscular Dystrophy Association, n.d.)

Standard-of-care-treatments: 
Current standard-of-care treatments for FSHD focus on managing symptoms and therapy due to no effective developed medication. Physical therapy and pain management through anti-inflammatory drugs are available to help improve mobility, muscular functions and reduce pain, however results vary depending on the stage of the disease. Scapulothoracic fusion surgery can benefit patients by improving shoulder related movements. Since conditions vary between individuals, respiratory, opticial, and hearing functions are carefully monitored as certain FSHD conditions may impact pulmonary, retinal vascular diseases, and hearing complications. While clinical trials have identified the pathology and etiology of the disease, the effective treatment remains unknown.(Aguirre et al., 2023)

Disease progression and prognosis: 
FSHD progresses slowly in individuals, often with symptoms appearing from early twenties and further developing over time. The rate and severity of FSHD is dependent on the individual and varies. As researched above, the progression of FSHD is shown within the severity of symptoms, however these changes do not affect life expectancy and a majority of patients maintain a normal lifespan. In patients with FSHD, weakness and loss of muscle mass occurs over time, commonly affecting facial muscles first and later spreading to other muscle groups causing increased axillary fat and arm muscle wasting. In severe cases where individuals are unable to maintain mobility due to muscle weakness, 20% will eventually require a wheelchair. (Muscular Dystrophy Association, n.d.)

Biological mechanisms: 
FSHD is the result of a regulatory mutation caused by the deletion of D4Z4 repeats on chromosome 4q. This deletion causes the transcription and overexpression of the DUX4 gene which is highly toxic. The overactivation of DUX4 instead of suppression in skeletal muscle, causes atrophy and protein degradation, which leads to irregular muscle function.This mechanism is hereditary, and is classified as an autosomal dominant trait. Potential therapies aim to reduce DUX4 expression or block its harmful effects on skeletal muscle cells. DUX4 target gene biomarkers in FSHD affected muscle could indicate levels of limb fat fraction. However, researchers continue investigating dependable biomarkers to monitor FSDH disease and optimal treatments. (Banerji et al., 2023; Hamel & Tawil, 2018)

Clinical trials and next generation therapies: 
Clinical trials for FSHD are testing therapies that target causes for FSHD. Currently a clinical study on phase three has been reviewing the impact of the drug Del-brax, an investigational RNA therapy designed to reduce DUX4 expression, has shown positive biomarker results, highlighting a decrease in KHDC1L, a biomarker linked the DUX4 overexpression, and a decrease in creatine kinase, a biomarker resulting in in muscle damage. While treatments have not been developed, these results indicate potential areas of treatment development. Future trials and testing will be required to find treatments for FSHD. Current and future testing should continue to identify reliable biomarkers and determine whether these therapies improve quality of life and mobility. (FSHD Society, 2026)

Prevalence and Incidence: 
The prevalence of FSHD in individuals in the United States is 1 in 15,000, and worldwide 2.03 to 6.8 per 100,000 individuals. The incidence of FSHD worldwide is 1 in 14,000 however the true estimate may be lower. (FSHD Canada, n.d.; Statland & Tawil, 2016)

Diagnosis: 
Diagnosis of FSHD is done through an evaluation by a neurologist or specialist in neuromuscular disorders. A patient is encouraged to seek testing after noticeable changes in their muscle strength or function. Identifiable traits are seen through muscle weakness in the face and shoulder blades. Testing is primarily completed through genetic testing to determine if the D4Z4 repeats have been shortened on chromosome 4q. In most cases. Genetic testing is able to confirm an FSHD diagnosis, however if inconclusive, blood tests, electromyography, or a muscle biopsy may be included in testing. (Muscular Dystrophy Association, n.d.)




## Data Set
The data set was based on a study cohort of 34 participants with FSHD who underwent functional studies, lower extremity MRI and bilateral biopsy of the tibialis anterior (TA) muscles. Participants were assessed at two timepoints, year one and two. Fat infiltration was calculated for each pixel by dividing the sum of the fat series intensity and water series signal intensities. AI- based segmentation was reviewed and evaluated by researchers for continued consistency of segmentation and to provide a final three dimensional label. Fiducial markers placed on the middle portion of the TA muscle were used to identify the muscle biopsy site. Fat infiltration was measured along the length of the muscle and expressed as a percentage. The provided csv data file features measurements such as DUX4 gene signatures, sex, age, dorsiflection strength, and clinical severity scores. This data source was from Wong et al. (2024).

## MRI Data Analysis Methodology

The MRI analysis was performed using data from one example FSHD subject measured at two visits, TP1 and TP2. Eight cross-sectional locations along the tibialis anterior (TA) muscle were sampled at each visit, giving 16 MRI sections total. For each sampled location, a water image, fat image, and tibialis anterior segmentation mask were provided. 

The water, fat, and mask files were loaded into Python using NumPy. The dimensions of the three arrays were checked to confirm that they represented the same MRI slice. The TA segmentation mask was binary, with pixels inside the tibialis anterior identified as 1 and pixels outside the muscle identified as 0. The mask was overlaid on both the water and fat images to visually confirm the location of the TA muscle.

Fat fraction was calculated at each valid pixel using:

**Fat Fraction (%) = 100 × Fat / (Water + Fat)**

Pixels where the combined water and fat signal was equal to zero were excluded to avoid division by zero. Fat fraction was first calculated for the whole MRI image and then separately for pixels located inside the TA segmentation mask. This allowed the analysis to focus specifically on the tibialis anterior muscle. The course methodology specifically uses paired water and fat images to calculate fat percentage in this way. 

The fat-fraction distribution inside the TA was visualized using a histogram. For each MRI slice, the mean fat fraction, median fat fraction, standard deviation, and number of valid TA pixels were calculated. These statistics were used to summarize the amount and variability of fat infiltration within the muscle.

The same analysis was repeated for all eight sampled locations at TP1 and all eight locations at TP2. Mean TA fat fraction was plotted against percent along the muscle so that the spatial pattern of fat infiltration could be compared between the two visits. The calculated measurements were also saved to a CSV file for later analysis. 

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Find the Module 2 project folder
base_folder = Path.cwd()

# If the notebook is opened from the GitHub repository folder,
# look inside the BME_2315 Module 2 folder
if not (base_folder / "Module2_FSHD_Dataset.csv").exists():
    possible_folder = base_folder / "BME_2315 Module 2"

    if (possible_folder / "Module2_FSHD_Dataset.csv").exists():
        base_folder = possible_folder

print("Project folder:")
print(base_folder)

# Load the main FSHD dataset
df = pd.read_csv(base_folder / "Module2_FSHD_Dataset.csv")

print("\nMain FSHD dataset shape:")
print(df.shape)

# Path to the MRI data
mri_folder = (
    base_folder
    / "FSHD_MRI_data"
    / "FSHD_MRI_data"
    / "FSHD_TA_MRI_data_0020_TP1_TP2"
)

# Load the MRI slice information
slice_info = pd.read_csv(mri_folder / "slice_info.csv")

print("\nMRI dataset shape:")
print(slice_info.shape)

print("\nMRI columns:")
print(slice_info.columns)

print("\nFirst five MRI rows:")
print(slice_info.head())
```

```python
# Select the first MRI slice
first_slice = slice_info.iloc[0]

# Load the matching water, fat, and TA mask files
water_image = np.load(mri_folder / first_slice["water_file"])
fat_image = np.load(mri_folder / first_slice["fat_file"])
ta_mask = np.load(mri_folder / first_slice["mask_file"])

print("Loaded MRI slice:")
print("Subject:", first_slice["subject"])
print("Timepoint:", first_slice["timepoint"])
print("Slice ID:", first_slice["slice_id"])

# Check that all three arrays have matching dimensions
print("\nArray dimensions:")
print("Water image:", water_image.shape)
print("Fat image:", fat_image.shape)
print("TA mask:", ta_mask.shape)

if water_image.shape == fat_image.shape == ta_mask.shape:
    print("All three arrays have matching dimensions.")
else:
    print("WARNING: Array dimensions do not match.")

# Check the values in the mask
print("\nUnique TA mask values:")
print(np.unique(ta_mask))

# Show the water MRI
plt.figure()
plt.imshow(water_image, cmap="gray")
plt.title("TP1 Slice 1 - Water MRI")
plt.xlabel("Pixel column")
plt.ylabel("Pixel row")
plt.show()

# Show the fat MRI
plt.figure()
plt.imshow(fat_image, cmap="gray")
plt.title("TP1 Slice 1 - Fat MRI")
plt.xlabel("Pixel column")
plt.ylabel("Pixel row")
plt.show()

# Show the TA mask
plt.figure()
plt.imshow(ta_mask, cmap="gray")
plt.title("TP1 Slice 1 - Tibialis Anterior Mask")
plt.xlabel("Pixel column")
plt.ylabel("Pixel row")
plt.show()

# Overlay the TA mask on the water MRI
plt.figure()
plt.imshow(water_image, cmap="gray")
plt.contour(ta_mask, levels=[0.5])
plt.title("TP1 Slice 1 - Water MRI with TA Mask")
plt.xlabel("Pixel column")
plt.ylabel("Pixel row")
plt.show()

# Overlay the TA mask on the fat MRI
plt.figure()
plt.imshow(fat_image, cmap="gray")
plt.contour(ta_mask, levels=[0.5])
plt.title("TP1 Slice 1 - Fat MRI with TA Mask")
plt.xlabel("Pixel column")
plt.ylabel("Pixel row")
plt.show()
```

```python
# Calculate total MRI signal
total_signal = water_image + fat_image

# Only use pixels where there is MRI signal
valid_pixels = total_signal > 0

# Calculate fat fraction
fat_fraction = np.zeros_like(fat_image, dtype=float)

fat_fraction[valid_pixels] = (
    fat_image[valid_pixels] / total_signal[valid_pixels]
) * 100

# Select valid pixels inside the tibialis anterior
ta_region = (ta_mask == 1) & valid_pixels
ta_fat_fraction = fat_fraction[ta_region]

# Calculate summary statistics
whole_image_mean = np.mean(fat_fraction[valid_pixels])

ta_mean = np.mean(ta_fat_fraction)
ta_median = np.median(ta_fat_fraction)
ta_std = np.std(ta_fat_fraction)
ta_pixel_count = len(ta_fat_fraction)

print("Fat Fraction Summary:")
print("Whole-image mean fat fraction:", whole_image_mean, "%")
print("TA mean fat fraction:", ta_mean, "%")
print("TA median fat fraction:", ta_median, "%")
print("TA standard deviation:", ta_std, "%")
print("Number of TA pixels:", ta_pixel_count)

# Show whole-image fat fraction map
plt.figure()
plt.imshow(fat_fraction, vmin=0, vmax=100)
plt.colorbar(label="Fat Fraction (%)")
plt.contour(ta_mask, levels=[0.5])
plt.title("TP1 Slice 1 - Fat Fraction Map with TA Boundary")
plt.xlabel("Pixel column")
plt.ylabel("Pixel row")
plt.show()

# Show TA-only fat fraction map
ta_only_map = np.full_like(fat_fraction, np.nan, dtype=float)
ta_only_map[ta_region] = fat_fraction[ta_region]

plt.figure()
plt.imshow(ta_only_map, vmin=0, vmax=100)
plt.colorbar(label="Fat Fraction (%)")
plt.title("TP1 Slice 1 - Tibialis Anterior Fat Fraction")
plt.xlabel("Pixel column")
plt.ylabel("Pixel row")
plt.show()

# Show TA fat fraction distribution
plt.figure()
plt.hist(ta_fat_fraction, bins=20)
plt.title("TP1 Slice 1 - TA Fat Fraction Distribution")
plt.xlabel("Fat Fraction (%)")
plt.ylabel("Number of Pixels")
plt.show()
```

```python
# Analyze all 16 MRI slices
results = []

for _, row in slice_info.iterrows():

    # Load the matching files
    water = np.load(mri_folder / row["water_file"])
    fat = np.load(mri_folder / row["fat_file"])
    mask = np.load(mri_folder / row["mask_file"])

    # Calculate fat fraction
    total_signal = water + fat
    valid_pixels = total_signal > 0

    slice_fat_fraction = np.zeros_like(fat, dtype=float)

    slice_fat_fraction[valid_pixels] = (
        fat[valid_pixels] / total_signal[valid_pixels]
    ) * 100

    # Select valid pixels inside the TA
    ta_region = (mask == 1) & valid_pixels
    ta_values = slice_fat_fraction[ta_region]

    # Save results from this slice
    results.append({
        "subject": row["subject"],
        "timepoint": row["timepoint"],
        "slice_id": row["slice_id"],
        "percent_along_muscle": row["percent_along_muscle"],
        "distance_from_muscle_start_mm": row["distance_from_muscle_start_mm"],
        "ta_mean_ff": np.mean(ta_values),
        "ta_median_ff": np.median(ta_values),
        "ta_std_ff": np.std(ta_values),
        "ta_pixel_count": len(ta_values)
    })

# Turn results into a DataFrame
results_df = pd.DataFrame(results)

print("MRI Fat Fraction Results:")
print(results_df)

# Save results to CSV
results_df.to_csv(
    base_folder / "mri_fat_fraction_results.csv",
    index=False
)

print("\nSaved: mri_fat_fraction_results.csv")

# Summarize by timepoint
timepoint_summary = results_df.groupby("timepoint")["ta_mean_ff"].agg(
    ["mean", "median", "std", "count"]
)

print("\nTA Fat Fraction Summary by Timepoint:")
print(timepoint_summary)

# Separate TP1 and TP2
tp1 = results_df[
    results_df["timepoint"] == "TP1"
].sort_values("percent_along_muscle")

tp2 = results_df[
    results_df["timepoint"] == "TP2"
].sort_values("percent_along_muscle")

# Plot mean TA fat fraction across muscle locations
plt.figure()

plt.plot(
    tp1["percent_along_muscle"],
    tp1["ta_mean_ff"],
    marker="o",
    label="TP1"
)

plt.plot(
    tp2["percent_along_muscle"],
    tp2["ta_mean_ff"],
    marker="o",
    label="TP2"
)

plt.title("TA Fat Fraction Across Muscle Locations")
plt.xlabel("Percent Along Muscle (%)")
plt.ylabel("Mean TA Fat Fraction (%)")
plt.legend()
plt.show()
```

### MRI Results Interpretation

Across the eight sampled locations, the average TA fat fraction increased from approximately **42.26% at TP1** to **54.41% at TP2**, an increase of about **12.15 percentage points**.

TP2 had a higher mean TA fat fraction than TP1 at most sampled locations. Fat fraction also varied substantially depending on the location along the tibialis anterior muscle. This shows why analyzing multiple locations along the muscle is more informative than using only one MRI slice.

These results describe one example subject, so they should not be interpreted as representing all patients with FSHD.

## Interpolation Analysis:

For these locations, we interpolated the TA fat fraction using linear interpolation between sections measured from the MRI. Location along the muscle was defined by percent along the muscle.

The measured MRI slices for the fat fraction were at 10%, 20%, 30%, 40%, 50%, 60%, 70% and 80% of muscle length. Linear interpolation was used for the fat fraction values at positions halfway between the measurements: 15%, 25%, 35%, 45%, 55%, 65% and 75%.

The interpolated values are an estimate and not a measurement since an MRI section was not actually taken at that estimated location. We assume that the fat fraction varies approximately linearly between two adjacent sampled locations.

All the estimated locations were between 10% and 80% of the muscle, so the analysis is based on interpolation not extrapolation. Potential sources of error include variation in fat fraction from MRI slice to slice, tibialis anterior segmentation, variation in MRI signal and the assumption of linear change between sampled locations.

```python
# Locations halfway between the measured MRI sections
interp_locations = np.array([15, 25, 35, 45, 55, 65, 75])

interpolation_results = []

plt.figure(figsize=(8, 5))

# Interpolate TP1 and TP2 separately
for timepoint in ["TP1", "TP2"]:

    data = results_df[
        results_df["timepoint"] == timepoint
    ].sort_values("percent_along_muscle")

    measured_locations = data["percent_along_muscle"].to_numpy()
    measured_ff = data["ta_mean_ff"].to_numpy()

    # Linear interpolation between measured locations
    interpolated_ff = np.interp(
        interp_locations,
        measured_locations,
        measured_ff
    )

    # Save the interpolated values
    for location, value in zip(interp_locations, interpolated_ff):
        interpolation_results.append({
            "timepoint": timepoint,
            "percent_along_muscle": location,
            "interpolated_ta_fat_fraction": value
        })

    # Plot measured values
    plt.plot(
        measured_locations,
        measured_ff,
        marker="o",
        label=f"{timepoint} measured"
    )

    # Plot interpolated values
    plt.scatter(
        interp_locations,
        interpolated_ff,
        marker="x",
        s=70,
        label=f"{timepoint} interpolated"
    )

plt.title("Measured and Interpolated TA Fat Fraction")
plt.xlabel("Percent Along Muscle (%)")
plt.ylabel("Mean TA Fat Fraction (%)")
plt.legend()
plt.tight_layout()
plt.show()

# Put the estimates into a DataFrame
interpolation_df = pd.DataFrame(interpolation_results)

print("Interpolated TA Fat Fraction Values:")
print(interpolation_df)

# Save the results
interpolation_df.to_csv(
    base_folder / "mri_interpolated_fat_fraction.csv",
    index=False
)

print("\nSaved: mri_interpolated_fat_fraction.csv")
```

# Interpolation Results

Seven additional estimates of TA fat fraction between the measured values at each MRI scan site were obtained through linear interpolation. These estimates of TA fat fraction varied with the pattern of the measured values.

In general, TP2 appeared to have a higher estimated fat fraction than TP1, and more so at the mid-point of the tibialis anterior. For example, at 45% of the total length the fat fraction was approximately 25.4% and 46.3% for TP1 and TP2 respectively.

These values are all approximations, not direct MRI measurements. The interpolation uses fat fraction values linearly between directly measured sites. Each interpolated measurement point was within the measured sites from 10% (muscle origin) to 80% (muscle end), so no extrapolated points are used.

## Biomarker Comparison: MRI Fat Fraction vs. STIR

MRI fat fraction and quantitative STIR were compared as two imaging biomarkers of FSHD progression. As only baseline and Year 1 STIR was available, both biomarkers were compared over the same baseline to Year 1 interval.

The tibialis anterior for each person was averaged for both the left and right sides so that I had one value per participant per biomarker per visit. From this I calculated change from baseline to Year 1. We are able to compare the two biomarkers using the same participants and same time interval.

```python
# Compare MRI fat fraction and STIR from baseline to Year 1

biomarker_df = pd.DataFrame()

biomarker_df["subject_id"] = df["subject_id"]

# Average left and right fat fraction values for each participant
biomarker_df["ff_baseline"] = df[
    ["ff_baseline_R", "ff_baseline_L"]
].mean(axis=1)

biomarker_df["ff_y1"] = df[
    ["ff_y1_R", "ff_y1_L"]
].mean(axis=1)

# Average left and right STIR values for each participant
biomarker_df["stir_baseline"] = df[
    ["stir_quant_baseline_R", "stir_quant_baseline_L"]
].mean(axis=1)

biomarker_df["stir_y1"] = df[
    ["stir_quant_y1_R", "stir_quant_y1_L"]
].mean(axis=1)

# Calculate change from baseline to Year 1
biomarker_df["ff_change"] = (
    biomarker_df["ff_y1"] - biomarker_df["ff_baseline"]
)

biomarker_df["stir_change"] = (
    biomarker_df["stir_y1"] - biomarker_df["stir_baseline"]
)

# Keep participants with both baseline and Year 1 data
comparison_df = biomarker_df.dropna(
    subset=[
        "ff_baseline",
        "ff_y1",
        "stir_baseline",
        "stir_y1"
    ]
).copy()

print("Number of participants with complete data:")
print(len(comparison_df))

print("\nFat Fraction Summary:")
print(
    comparison_df[
        ["ff_baseline", "ff_y1", "ff_change"]
    ].describe()
)

print("\nSTIR Summary:")
print(
    comparison_df[
        ["stir_baseline", "stir_y1", "stir_change"]
    ].describe()
)

# Calculate average changes
mean_ff_change = comparison_df["ff_change"].mean()
mean_stir_change = comparison_df["stir_change"].mean()

median_ff_change = comparison_df["ff_change"].median()
median_stir_change = comparison_df["stir_change"].median()

print("\nAverage change from baseline to Year 1:")
print("Fat fraction:", mean_ff_change)
print("STIR:", mean_stir_change)

print("\nMedian change from baseline to Year 1:")
print("Fat fraction:", median_ff_change)
print("STIR:", median_stir_change)

# Plot individual participant changes in fat fraction
plt.figure(figsize=(7, 5))

for _, row in comparison_df.iterrows():
    plt.plot(
        ["Baseline", "Year 1"],
        [row["ff_baseline"], row["ff_y1"]],
        marker="o",
        alpha=0.5
    )

plt.title("MRI Fat Fraction: Baseline to Year 1")
plt.xlabel("Visit")
plt.ylabel("Mean TA Fat Fraction")
plt.tight_layout()
plt.show()

# Plot individual participant changes in STIR
plt.figure(figsize=(7, 5))

for _, row in comparison_df.iterrows():
    plt.plot(
        ["Baseline", "Year 1"],
        [row["stir_baseline"], row["stir_y1"]],
        marker="o",
        alpha=0.5
    )

plt.title("Quantitative STIR: Baseline to Year 1")
plt.xlabel("Visit")
plt.ylabel("Mean TA STIR Measurement")
plt.tight_layout()
plt.show()

# Save comparison results
comparison_df.to_csv(
    base_folder / "fat_fraction_vs_stir_comparison.csv",
    index=False
)

print("\nSaved: fat_fraction_vs_stir_comparison.csv")
```

# Comparison of Biomarker Results

Nineteen subjects had quantitative STIR at baseline and Year 1 in addition to MRI fat fraction data.

Mean tibialis anterior fat fraction values grew from 25.02 (average of baseline) to 26.35 (average at Year 1) representing an average change of 1.33. Median change was 1.13. The change in fat fraction was fairly consistent across the participants.

Quantitative STIR was observed to be more variable across individuals. The average STIR score was 14.49 before treatment and 14.31 at Year 1, an average change of roughly -0.18. There were some large increases and large decreases among the individuals.

Since the units of change are different between the fat fraction and STIR measurements, the raw values cannot be directly compared in terms of magnitude of change. The fat-fraction values were more consistent across participants, while STIR was highly variable. The MRI fat-fraction value was more consistent within this dataset, indicating that it may be the better biomarker in this case.

```python
# Compare consistency across the two biomarkers using standardized response mean
ff_srm = (
    comparison_df["ff_change"].mean()
    / comparison_df["ff_change"].std()
)

stir_srm = (
    comparison_df["stir_change"].mean()
    / comparison_df["stir_change"].std()
)

print("Standardized Response Mean:")
print("Fat fraction:", ff_srm)
print("STIR:", stir_srm)
```

The SRM was in the order of 0.87 for fat fraction and -0.009 for STIR. This further confirms that the fat fraction had greater responsiveness over baseline to Year 1, and that STIR was not showing consistent direction of change relative to its variability.

## Conclusions:

These results indicate that MRI fat fraction is a promising biomarker for tracking FSHD progress. For the example MRI subject, tibialis anterior fat fraction was increased at TP2 relative to TP1 at most of the sampled muscle sites, and the shape of this increase was maintained when interpolated between measured sites.

In the NEXUS MRI study, mean MRI fat fraction in 19 participants from baseline to Year 1 was more stable and increased, with about 1.33 change in the mean between baseline and 1-year follow-up, compared with the more variable mean change for quantitative STIR, which changed very little on average and showed a much greater variability. The SRM was 0.87 for fat fraction and -0.009 for STIR.

From these data, the MRI fat fraction seems to be the more promising imaging biomarker of the two to potentially use as a secondary exploratory endpoint in an early FSHD clinical trial. But bigger studies with more patients and more follow-up times would be required before using fat fraction to measure response to treatment. And the researchers would need to know if changes in MRI fat fraction were associated with improvements in muscle strength, physical function, or quality of life.

You will also need to make decisions about possible ethical issues in choosing a clinical-trial endpoint. The use of MRI as an objective index of muscle structure may be more costly or less feasible than more standard clinical measures of such structure. MRI-based biomarkers should be evaluated in the context of other measures that indicate whether changes are meaningful to patients, such as mobility, strength, and functional capacity.

## Limitations and Future Work:

A significant limitation of this project was the relatively small sample size, which likely contributed to the lack of statistically significant findings. The dataset was based off a study cohort of 34 participants with facioscapulohumeral muscular dystrophy (FSHD). If more time was allotted to this project, further analysis could be conducted by evaluating another biomarker such as the DUX4 transcriptome signatures and comparing their efficacy against the MRI fat fraction and muscle strength. For future analysis, our research would focus on evaluating the clinical advancement of epigenetic and RNA target therapies developed to suppress the toxic DUX4 overexpression. Given that an estimated 20% of individuals living with FSHD eventually require a wheelchair for mobility, further research in this area allows for developing potential enhancements in clinical trials.
